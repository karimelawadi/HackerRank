```python
a = int(input())
b = int(input())
m = int(input())
print(pow(a,b))
print(pow(a,b,m))
```

Python Math Mod Divmod

```python
a = int(input())
b = int(input())
print(a//b)
print(a%b)
print (divmod(a,b))
```

Python Math Find Angle MBC

```python
import math
import cmath

AB = int(input())
BC = int(input())
hypotenuse = cmath.sqrt(AB**2 + BC**2)
angle = cmath.asin(AB/hypotenuse)
print(str(math.floor(angle.real*180/math.pi + 0.5)) + chr(176))
```
