Python Collections collections.Counter()

```python
from collections import Counter

X = int(input())
shoes = Counter(map(int, input().split()))

N = int(input())
earning = 0

for _ in range(N):
    size, price = map(int, input().split())
    if shoes[size] > 0:
        earning += price
        shoes[size] -= 1

print(earning)
```
