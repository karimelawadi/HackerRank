Python Regex and Parsing Matrix Script

```python
import math
import os
import random
import re
import sys




first_multiple_input = input().rstrip().split()

n = int(first_multiple_input[0])

m = int(first_multiple_input[1])

matrix = []

for _ in range(n):
    matrix_item = input()
    matrix.append(matrix_item)
decoded = ''
for col in range(m):
    for row in range(n):
        decoded += matrix[row][col]

result = re.sub(r'(?<=[A-Za-z0-9])[^A-Za-z0-9]+(?=[A-Za-z0-9])', ' ', decoded)

print(result)


#------------------------------------------------------------------------------------------------


import re

start= input('row : colomn\t').split()
rows= int(start[0])
coloms= int(start[1])


matrix = [(lambda row : row)  (input())  for _ in range(rows)]

s=''    
for Colom in range (coloms):
    for Row in range (rows):
        s+= matrix[Row][Colom]
    


result = re.sub(r'(?<=[A-Za-z0-9])[^A-Za-z0-9]+(?=[A-Za-z0-9])', ' ', s)

print(result)

```
