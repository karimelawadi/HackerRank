Python Errors and Exceptions Exceptions

```python
t = int(input())
for _ in range(t):
    try:
        a, b = input().split()
        print(int(a) // int(b))
    except (ZeroDivisionError, ValueError) as e:
        print("Error Code:", e)
```

Python Errors and Exceptions Incorrect Regex

```python
import re

for _ in range(int(input())):
    try:
        re.compile(input())
        print("True")
    except re.error:
        print("False")
```
