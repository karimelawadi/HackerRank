Python Sets Introduction to Sets

```python
def average(array):
    
    distinct = set(array)
    return round(sum(distinct) / len(distinct), 3)
if __name__ == '__main__':
    n = int(input())
    arr = list(map(int, input().split()))
    result = average(arr)
    print(result)
```

Python Sets Set .discard(), .remove() & .pop()

```python
n = int(input())
s = set(map(int, input().split()))
N = int(input())

for _ in range(N):
    command = input().split()
    op = command[0]
    
    if op == 'pop':
        s.pop()
    elif op == 'remove':
        s.remove(int(command[1]))
    elif op == 'discard':
        s.discard(int(command[1]))

print(sum(s))
```

Python Sets Set Mutations

```python
n = int(input())
A = set(map(int, input().split()))

num_ops = int(input())
for _ in range(num_ops):
    op_name, _ = input().split()
    other_set = set(map(int, input().split()))
    
    if op_name == "update":
        A.update(other_set)
    elif op_name == "intersection_update":
        A.intersection_update(other_set)
    elif op_name == "difference_update":
        A.difference_update(other_set)
    elif op_name == "symmetric_difference_update":
        A.symmetric_difference_update(other_set)

print(sum(A))
```

Python Sets The Captain's Room

```python
from collections import Counter

K = int(input())
rooms = list(map(int, input().split()))

counts = Counter(rooms)
for room, count in counts.items():
    if count != K:
        print(room)
        break
```

```python
T = int(input())

for _ in range(T):
    n = int(input())
    A = set(input().split())
    
    m = int(input())
    B = set(input().split())
    
    print(A.issubset(B))
```

Python Sets Check Strict Superset

```python
A = set(input().split())
n = int(input())

result = True
for _ in range(n):
    other = set(input().split())
    if not A.issuperset(other) or A == other:
        result = False

print(result)
```
