Python Strings Find a string

```python
def count_substring(string, sub_string):
    count = 0
    for section in range(0, len(string) - len(sub_string) + 1):
        result = string[section : section + len(sub_string)]
        if result == sub_string:
            count += 1
    return count

if __name__ == '__main__':
    string = input().strip()
    sub_string = input().strip()
    count = count_substring(string, sub_string)
    print(count)

```

Python Strings sWAP cASE

```python
def swap_case(s):
    result =''
    for cracter in s:
        if cracter == cracter.upper() :
             result = result +cracter.lower()
        else :result = result +cracter.upper()
    return result   
        

if __name__ == '__main__':
    s = input()
    result = swap_case(s)
    print(result)

```

Python Strings String Split and join

```python
def split_and_join(line):
    line=line.split()
    line=('-'.join(line))
    return line

if __name__ == '__main__':
    line = input()
    result = split_and_join(line)
    print(result)
```

Python Strings Mutations

```python
def mutate_string(string, position, character):
    list(string)
    return string[:position] + character + string[position+1:]


if __name__ == '__main__':
    s = input()
    i, c = input().split()
    s_new = mutate_string(s, int(i), c)
    print(s_new)
```

Python Strings Text Wrap

```python
import textwrap

def wrap(string, max_width):
   parts = []
   for caracter in range(0,len(string),max_width):
      parts.append(string[caracter:max_width+caracter])
   return '\n'.join(parts)    

if __name__ == '__main__':
    string, max_width = input(), int(input())
    result = wrap(string, max_width)
    print(result)
```

Python Strings String Capitalize !

```python
import math
import os
import random
import re
import sys


def solve(s):
    words = s.split(' ')
    capitalized = []
    for w in words:
        capitalized.append(w.capitalize())
    return ' '.join(capitalized)


if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    s = input()

    result = solve(s)

    fptr.write(result + '\n')

    fptr.close()
```

Python Strings Merge the Tools!

```python
def merge_the_tools(string, k):
    result = []
    for i in range(0, len(string), k):
        chunk = string[i:i+k]
        new_string = ''
        for char in chunk:
            if char not in new_string:
                new_string += char
        result.append(new_string)
    print(result)

if __name__ == '__main__':
    string, k = input(), int(input())
    merge_the_tools(string, k)


```
